```python
# importing materials
import numpy as np
import matplotlib.pyplot as plt
import statistics as st
```

```python
# importing data from sst and pvalb data sets on github
url_sst = "https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_sst.csv"
url_pvalb = "https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_pvalb.csv"
v_rest1 = np.loadtxt(url_sst)
v_rest2 = np.loadtxt(url_pvalb)
```

```python
# concentrating the two variables into a matrix called v_rest
v_rest = np.array([v_rest1, v_rest2])
print(v_rest.shape)
```

```python
# printing the size of v_rest
print(v_rest.size)
```

```python
# printing the first row of v_rest
print(v_rest[0])
```

```python
# printing the sum of the values in the third column of v_rest
print(v_rest[:,2].sum())
```

```python
# printing the second to last value in the second row of v_rest
print(v_rest[:1,-1])
```

```python
# making a histogram of the values in the first row of v_rest, to see how frequent each resting potential was
plt.hist(v_rest[0])
plt.xlabel('resting potential (mV)')
plt.ylabel('frequency')
```

```python
# calculating the difference between the two means between the two rows
v_rest_mean_sst = np.mean(v_rest[0])
v_rest_mean_pvalb = np.mean(v_rest[1])
print(v_rest_mean_sst)
print(v_rest_mean_pvalb)
```

The Pvalb mean is more negative than the sst mean. This indicates that on average the resting potential in the Pvalb neurons is more negative, it is thus farther from depolarization at rest. However, the two averages are very similar.

```python
# importing statistics materials
import scipy.stats as stats
```

```python
# using the stats.ttest function to see if my results were statistically significant.
stats.ttest_ind(v_rest[0], v_rest[1])
```

According to the t-test results, my p-value was about 7.16x10^-9. Since this number is smaller than 0.05, it indicates that my results were statistically significant.
