```python
# Making Panda Variables from .csv files
url1 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_sst.csv'
pandas_data1 = pd.read_csv(url1)
url2 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_pvalb.csv'
pandas_data2 = pd.read_csv(url2)
```

```python
# Converting pandas data into a list variable
data1 = pandas_data1.values.tolist()
data2 = pandas_data2.values.tolist()
```

```python
# preventing brackets around every individual element
data1 = np.squeeze(data1)
data2 = np.squeeze(data2)
```

```python
print(data1)
```

```python
# Printing 3rd, 4th, 5th, and 6th elements of data2
print(data2[2:6])

```

```python
# Printing second-to-last and last elements of data1
print(data1[-2:])
```

```python
# Printing the type of data2
print(type(data2))
```

```python
# Making a histogram of data1
plt.hist(data1)
```

```python
# labeling x and y axis
plt.hist(data1)
plt.xlabel('resting potential (mV)')
plt.ylabel('frequency')
```

```python
# histogram of data2
plt.hist(data2)
plt.xlabel('resting potential (mV)')
plt.ylabel('frequency')
```
