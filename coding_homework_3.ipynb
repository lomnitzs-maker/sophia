```python
# importing necessary materials
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats
```

```python
# importing data as numpy arrays
url_sst = "https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_sst.csv"
url_pvalb = "https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_pvalb.csv"
v_rest_sst = np.loadtxt(url_sst)
v_rest_pvalb = np.loadtxt(url_pvalb)
```

```python
# printing the means of both data sets, preforming a t-test to test the null hypothesis that the means of v_rest_sst and v_rest_pvalb come from different distributions
# plotting a histogram of v_rest in one subplot and a histogram of v_rest_pvalb if the p-value of the t-test is < 0.05
v_rest_sst_avg = np.mean(v_rest_sst)
v_rest_pvalb_avg = np.mean(v_rest_pvalb)
print(v_rest_sst_avg)
print(v_rest_pvalb_avg)

t_stat, p_val = stats.ttest_ind(v_rest_sst, v_rest_pvalb)
print("t=", t_stat)
print("p=", p_val)
alpha = 0.05
if p_val <0.05:
  print("Reject the null hypothesis")
else:
  print("Fail to reject the null hypothesis")

fig, ax = plt.subplots(1,2,figsize=(15,5))
ax[0].hist(v_rest_sst)
ax[0].set_ylabel('frequency')
ax[0].set_xlabel('resting potential')
ax[0].set_title('Sst')
if p_val < 0.05:
  ax[1].hist(v_rest_pvalb)
  ax[1].set_ylabel('frequency')
  ax[1].set_xlabel('resting potential')
  ax[1].set_title('Pvalb')
plt.show()




```
