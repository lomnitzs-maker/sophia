```python
# importing relevant modules
import numpy as np
import matplotlib.pyplot as plt
```

```python
# importing data
acc_hits = np.loadtxt('https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/acc_hits_hw5.csv', delimiter = ',')
acc_fa = np.loadtxt('https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/acc_fa_hw5.csv', delimiter = ',')
```

```python
# printing the shape of both my arrays
print(acc_hits.shape)
print(acc_fa.shape)
```

```python
# creating a time variable that has the same length as acc_hits and acc_fa
time = np.linspace(0,4,8001)
```

```python
# creating an expression to determine the sampling rate of the data
sampling_rate = (8001-1)/4
```

```python
fig, ax = plt.subplots(1,2,figsize=(15,5))
ax[0].plot(time,acc_hits[0,:])
ax[0].set_xlabel('time')
ax[0].set_ylabel('LFP')
ax[0].set_title('Hits')
ax[1].plot(time,acc_fa[0,:])
ax[1].set_xlabel('time')
ax[1].set_ylabel('LFP')
ax[1].set_title('False Alarms')
```

I think the dominat ossillation in this LFP is 140 Hz, which occurs during the first trial of the "Hits". It does differ from the first trial of false alarms, which has a dominant oscillation of about 23 Hz.

```python
# making a graph to compare LFPs between hits and false alarms, using ERPs to do so
erp_hits = np.mean(acc_hits, axis=0)
plt.plot(time,erp_hits, label='ERP hits')
plt.xlabel('Time')
plt.ylabel('ERP')
plt.title('ERP for acc_hits')
plt.axvline(x=2, color='black', label='Screen Touch')
plt.axhline(y=0, color='black')
plt.legend()
```

Yes, once the time hits two seconds, which is where the verticle black line is, there is an obvious change in activity. The activity goes from being relatively stable to having signifcantly larger peaks and troughs.

```python
# repeating step 10 for acc_fa
erp_fa = np.mean(acc_fa, axis=0)
plt.plot(time,erp_fa, label='ERP fa')
plt.xlabel('Time')
plt.ylabel('ERP')
plt.title('ERP for acc_fa')
plt.axvline(x=2, color='black', label='Screen Touch')
plt.axhline(y=0, color='black')
```

Yes, there are differences between ERPs in the hits and false alarm graphs. In the false alarm graph, the activity is consistent throughout the graph. The peaks and troughs are pretty consistent in size, the only thing is that they start off in the positive half of the graph, and then begin to migrate towards the negative half. But, they go back up again and then back down. Overall, the activity is a lot more consistent in the false alarm ERPs than the hits ERPs.
