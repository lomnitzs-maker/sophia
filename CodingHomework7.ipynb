```python
# importing necesary materials
import numpy as np
import matplotlib.pyplot as plt
```

```python
# uploading each file from the spikes csv data and then creating a variable for each set of data
urll1 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_left_trial1.csv'
urlr1 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_right_trial1.csv'
urll2 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_left_trial2.csv'
urlr2 = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_right_trial2.csv'

spikes_left_trial1 = np.loadtxt(urll1, delimiter = ',')
spikes_right_trial1 = np.loadtxt(urlr1, delimiter = ',')
spikes_left_trial2 = np.loadtxt(urll2, delimiter = ',')
spikes_right_trial2 = np.loadtxt(urll2, delimiter = ',')
```

```python
# printing the length of each array
print(len(spikes_left_trial1))
print(len(spikes_right_trial1))
print(len(spikes_left_trial2))
print(len(spikes_right_trial2))
```

```python
# determining the firing rate of the neuron during each trials
print(len(spikes_left_trial1)/30)
print(len(spikes_right_trial1)/30)
print(len(spikes_left_trial2)/30)
print(len(spikes_right_trial2)/30)
```

The neuron fires more before left turn trials, because it's firing rate is higher. The firing rate is 2.17 spikes per second,  whereas for right trial one the neuron has a firing rate of 1.5 spikes per second.

```python
# making two subplots to plot the spike train data from trial 1 in one subplot and trial 2 in the other subplot
fig, ax = plt.subplots(1,2,figsize=(15,5))
ax[0].plot(spikes_left_trial1,np.ones_like(spikes_left_trial1),'.')
ax[0].plot(spikes_right_trial1,1.5*np.ones_like(spikes_right_trial1), '.')
ax[0].set_xlim([0,30])
ax[0].set_ylim([0,2])
ax[0].set_ylabel('Trial Type')
ax[0].set_xlabel('time[s]')
ax[0].set_title('Trial 1')
ax[0].set_yticks([1, 1.5])
ax[0].set_yticklabels(['Left', 'Right'])

ax[1].plot(spikes_left_trial2,np.ones_like(spikes_left_trial2),'.')
ax[1].plot(spikes_right_trial2,1.5*np.ones_like(spikes_right_trial2),'.')

#ax[1].plot(spikes_left_trial2)
#ax[1].plot(spikes_right_trial2)
ax[1].set_xlim([0,30])
ax[1].set_ylim([0,2])
ax[1].set_ylabel('trial type')
ax[1].set_xlabel('time[s]')
ax[1].set_title('Trial 2')
ax[1].set_yticks([1, 1.5])
ax[1].set_yticklabels(['Left', 'Right'])
```

This neuron does not seem more bursty during either left or right turn trials. I would say in trial 1 the left is probably a little bit burstier while during the right turn trial it is more regular firing.

```python
# calculating inter-spike interval (ISI) arrays for each of the four spike arrays
ISIsLeft1 = np.diff(spikes_left_trial1)
ISIsRight1 = np.diff(spikes_right_trial1)
ISILeft2 = np.diff(spikes_left_trial2)
ISIRight2 = np.diff(spikes_right_trial2)
```

```python
# Making four histograms for each of my ISI arrays
fig, ax = plt.subplots(2,2, figsize =(15,5))
bins = np.arange(0,2.01,0.01)

ax[0,0].hist(ISIsLeft1, bins=bins)
ax[0,0].set_title('Left Trial 1')
ax[0,0].set_xlabel('ISI (s)')
ax[0,0].set_ylabel('Count')

ax[0,1].hist(ISILeft2, bins=bins)
ax[0,1].set_title('Left Trial 2')
ax[0,1].set_xlabel('ISI (s)')
ax[0,1].set_ylabel('Count')

ax[1,0].hist(ISIsRight1, bins=bins)
ax[1,0].set_title('Right Trial 1')
ax[1,0].set_xlabel('ISI (s)')
ax[1,0].set_ylabel('Count')

ax[1,1].hist(ISIRight2, bins=bins)
ax[1,1].set_title('Right Trial 2')
ax[1,1].set_xlabel('ISI (s)')
ax[1,1].set_ylabel('Count')

plt.tight_layout()

plt.show()
```

Again, during left trial 1, the neuron seems a little burstier in comparison to right trial 1. Left trial 2 and right trial 2 are exactly the same. Ultimatley these histograms don't leave me with any big takeways about the burtiness of the neuron during the left vs. right trials.

```python
bins = np.arange(0,31,1)
fr_left_trial1, _ = np.histogram(spikes_left_trial1, bins=bins)
fr_right_trial1, _ = np.histogram(spikes_right_trial1, bins=bins)
fr_left_trial2, _ = np.histogram(spikes_left_trial2, bins=bins)
fr_right_trial2, _ = np.histogram(spikes_right_trial2, bins=bins)
time = bins[:-1]
fig, ax = plt.subplots(1,2, figsize=(15,5))

#Trial 1
ax[0].plot(time,fr_left_trial1, color='red',label='Left')
ax[0].plot(time,fr_right_trial1,color='blue',label='Right')
ax[0].set_xlabel('Time (s)')
ax[0].set_ylabel('Firing rate (Hz)')
ax[0].set_title('Trial 1')
ax[0].legend()

#trial 2
ax[1].plot(time,fr_left_trial2,color='red',label='Left')
ax[1].plot(time,fr_right_trial2, color='blue', label='Right')
ax[1].set_xlabel('Time (s)')
ax[1].set_ylabel('Firing rate (Hz)')
ax[1].set_title('Trial 2')
ax[1].legend()

plt.tight_layout()
plt.show
```

Yes, during the left trial between 10-15 seconds the neuron is having a huge spikes while at this time for the 1st right trial there are not such large spikes. However, at the 30 second mark there was a large spike during the 1st right trial where there was no such spike during the left trial.
